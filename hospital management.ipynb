```python
import pandas as pd

original_data = pd.read_csv("data/Original_Dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", original_data.shape[0])
print("Columns:", original_data.shape[1])

original_data.head()
```

```python
print(original_data.columns.tolist())
original_data.info()
```

```python
# Create a copy so that original dataset remains safe
df_ml = original_data.copy()

# Replace missing values with empty text
df_ml = df_ml.fillna("")

print("Missing values handled successfully!")

print("\nRemaining missing values:")
print(df_ml.isnull().sum().sum())
```

```python
# Get all symptom columns
symptom_columns = [
    col for col in df_ml.columns
    if col.startswith("Symptom_")
]

# Combine all symptoms into one text column
df_ml["All_Symptoms"] = df_ml[symptom_columns].apply(
    lambda row: " ".join(
        str(symptom).strip().replace("_", " ")
        for symptom in row
        if str(symptom).strip() != ""
    ),
    axis=1
)

# Display the result
df_ml[["Disease", "All_Symptoms"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df_ml["All_Symptoms"])
y = df_ml["Disease"]

print("X shape:", X.shape)
print("Number of diseases:", y.nunique())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(X_train, y_train)

print("ML model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
print("Total records:", len(df_ml))

duplicates = df_ml.duplicated(
    subset=["Disease", "All_Symptoms"]
).sum()

print("Duplicate records:", duplicates)
```

```python
df_clean = df_ml.drop_duplicates(
    subset=["Disease", "All_Symptoms"]
).reset_index(drop=True)

print("Original records:", len(df_ml))
print("After removing duplicates:", len(df_clean))
```

```python
# Create TF-IDF features from the clean dataset

X_clean = vectorizer.fit_transform(df_clean["All_Symptoms"])
y_clean = df_clean["Disease"]

print("Clean X shape:", X_clean.shape)
print("Number of diseases:", y_clean.nunique())
```

```python
X_train_clean, X_test_clean, y_train_clean, y_test_clean = train_test_split(
    X_clean,
    y_clean,
    test_size=0.20,
    random_state=42,
    stratify=y_clean
)

print("Training records:", X_train_clean.shape[0])
print("Testing records:", X_test_clean.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression

clean_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

clean_model.fit(X_train_clean, y_train_clean)

print("Clean ML model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score

y_pred_clean = clean_model.predict(X_test_clean)

clean_accuracy = accuracy_score(
    y_test_clean,
    y_pred_clean
)

print("Clean Model Accuracy:", round(clean_accuracy * 100, 2), "%")
```

```python
def find_available_doctors(specialist, duty_date):
    
    query = """
        SELECT
            d.doctor_name,
            d.specialization,
            dep.department_name,
            ds.shift_name,
            ds.start_time,
            ds.end_time,
            d.room_number,
            ds.status
        FROM doctor_shifts ds
        JOIN doctors d
            ON ds.doctor_id = d.doctor_id
        JOIN departments dep
            ON d.department_id = dep.department_id
        WHERE d.specialization = ?
          AND ds.duty_date = ?
          AND ds.status = 'Available'
        ORDER BY ds.start_time
    """
    
    return pd.read_sql_query(
        query,
        conn,
        params=(specialist, duty_date)
    )
```

```python
def complete_recommendation(symptoms, duty_date):

    # Convert symptoms to text
    symptom_text = " ".join(
        symptom.strip().replace("_", " ")
        for symptom in symptoms
        if symptom.strip() != ""
    )

    # Convert symptoms into TF-IDF features
    symptom_vector = vectorizer.transform([symptom_text])

    # Predict disease
    predicted_disease = clean_model.predict(symptom_vector)[0]

    # Get disease description
    description = disease_to_description.get(
        predicted_disease,
        "Description not available"
    )

    # Get specialist
    specialist = disease_to_specialist.get(
        predicted_disease,
        "Specialist mapping not available"
    )

    # Find available doctors
    if specialist != "Specialist mapping not available":
        doctors = find_available_doctors(
            specialist,
            duty_date
        )
    else:
        doctors = pd.DataFrame()

    return predicted_disease, description, specialist, doctors
```

```python
symptom_columns = [
    col for col in original_data.columns
    if col.startswith("Symptom_")
]

df_ml = original_data.copy()

df_ml["All_Symptoms"] = df_ml[symptom_columns].fillna("").apply(
    lambda row: " ".join(
        str(x).strip().replace("_", " ")
        for x in row
        if str(x).strip() != ""
    ),
    axis=1
)

df_clean = df_ml.drop_duplicates(
    subset=["Disease", "All_Symptoms"]
).reset_index(drop=True)

print("Original records:", len(df_ml))
print("Clean records:", len(df_clean))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

vectorizer = TfidfVectorizer()

X_clean = vectorizer.fit_transform(df_clean["All_Symptoms"])
y_clean = df_clean["Disease"]

X_train_clean, X_test_clean, y_train_clean, y_test_clean = train_test_split(
    X_clean,
    y_clean,
    test_size=0.20,
    random_state=42,
    stratify=y_clean
)

clean_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

clean_model.fit(X_train_clean, y_train_clean)

print("Clean model trained successfully!")
print("Training records:", X_train_clean.shape[0])
print("Testing records:", X_test_clean.shape[0])
```

```python
# Load Disease → Specialist mapping
doctor_disease = pd.read_csv(
    "data/Doctor_Versus_Disease.csv",
    encoding="latin1"
)

doctor_disease.columns = ["Disease", "Specialist"]

doctor_disease["Disease"] = doctor_disease["Disease"].astype(str).str.strip()
doctor_disease["Specialist"] = (
    doctor_disease["Specialist"]
    .astype(str)
    .str.replace("\xa0", " ", regex=False)
    .str.strip()
)

disease_to_specialist = dict(
    zip(
        doctor_disease["Disease"],
        doctor_disease["Specialist"]
    )
)

# Load Disease → Description mapping
disease_description = pd.read_csv(
    "data/Disease_Description.csv",
    encoding="latin1"
)

disease_description.columns = ["Disease", "Description"]

disease_description["Disease"] = (
    disease_description["Disease"].astype(str).str.strip()
)

disease_to_description = dict(
    zip(
        disease_description["Disease"],
        disease_description["Description"]
    )
)

print("Disease → Specialist:", len(disease_to_specialist))
print("Disease → Description:", len(disease_to_description))
```

```python
def find_available_doctors(specialist, duty_date):

    query = """
        SELECT
            d.doctor_name,
            d.specialization,
            dep.department_name,
            ds.shift_name,
            ds.start_time,
            ds.end_time,
            d.room_number,
            ds.status
        FROM doctor_shifts ds
        JOIN doctors d
            ON ds.doctor_id = d.doctor_id
        JOIN departments dep
            ON d.department_id = dep.department_id
        WHERE d.specialization = ?
          AND ds.duty_date = ?
          AND ds.status = 'Available'
        ORDER BY ds.start_time
    """

    return pd.read_sql_query(
        query,
        conn,
        params=(specialist, duty_date)
    )

print("Doctor search function ready!")
```

```python
def complete_recommendation(symptoms, duty_date):

    # Convert symptoms into text
    symptom_text = " ".join(
        symptom.strip().replace("_", " ")
        for symptom in symptoms
        if symptom.strip() != ""
    )

    # Convert symptoms into TF-IDF features
    symptom_vector = vectorizer.transform([symptom_text])

    # Predict disease
    predicted_disease = clean_model.predict(symptom_vector)[0]

    # Get disease description
    description = disease_to_description.get(
        predicted_disease,
        "Description not available"
    )

    # Get specialist
    specialist = disease_to_specialist.get(
        predicted_disease,
        "Specialist mapping not available"
    )

    # Find available doctors
    if specialist != "Specialist mapping not available":
        doctors = find_available_doctors(
            specialist,
            duty_date
        )
    else:
        doctors = pd.DataFrame()

    return predicted_disease, description, specialist, doctors

print("Complete recommendation function ready!")
```

```python
symptoms = [
    "itching",
    "skin rash",
    "nodal skin eruptions"
]

predicted_disease, description, specialist, doctors = complete_recommendation(
    symptoms,
    "2026-08-24"
)

print("Predicted Disease:", predicted_disease)

print("\nDisease Description:")
print(description)

print("\nRecommended Specialist:", specialist)

print("\nAvailable Doctors:")
display(doctors)
```

```python
def get_patient_history(patient_id):

    query = """
        SELECT
            h.history_id,
            p.patient_name,
            d.doctor_name,
            h.visit_date,
            h.department,
            h.symptoms,
            h.notes
        FROM patient_history h
        JOIN patients p
            ON h.patient_id = p.patient_id
        JOIN doctors d
            ON h.doctor_id = d.doctor_id
        WHERE h.patient_id = ?
        ORDER BY h.visit_date DESC
    """

    return pd.read_sql_query(
        query,
        conn,
        params=(patient_id,)
    )

print("Patient history function ready!")
```
